```python
# ============================================================
# LOW LIGHT IMAGE ENHANCEMENT PROJECT (FULL READY CODE)
# GOOGLE COLAB VERSION
# U-NET + 50 EPOCHS + PSNR + SSIM + BRIGHTNESS + TESTING
# ============================================================

# ============================================================
# STEP 1 : INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install torchmetrics -q

# ============================================================
# STEP 2 : IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

from torchmetrics.functional import peak_signal_noise_ratio
from torchmetrics.functional import structural_similarity_index_measure

print("Libraries Imported Successfully")

# ============================================================
# STEP 3 : GOOGLE DRIVE MOUNT
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

# ============================================================
# STEP 4 : DATASET PATH
# ============================================================

LOW_LIGHT_PATH = "/content/drive/MyDrive/lol_dataset/our485/low"
NORMAL_LIGHT_PATH = "/content/drive/MyDrive/lol_dataset/our485/high"

TEST_LOW_PATH = "/content/drive/MyDrive/lol_dataset/eval15/low"
TEST_HIGH_PATH = "/content/drive/MyDrive/lol_dataset/eval15/high"

SAVE_MODEL_PATH = "/content/drive/MyDrive/best_low_light_model.pth"

print("Dataset Path Ready")

# ============================================================
# STEP 5 : CUSTOM DATASET transform
# ============================================================

class LowLightDataset(Dataset):
    def __init__(self, low_dir, high_dir):
        self.low_dir = low_dir
        self.high_dir = high_dir

        self.images = sorted(os.listdir(low_dir))

        self.transform = transforms.Compose([
            transforms.Resize((128, 128)),
            transforms.ToTensor()
        ])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        file_name = self.images[idx]

        low_path = os.path.join(self.low_dir, file_name)
        high_path = os.path.join(self.high_dir, file_name)

        low_img = Image.open(low_path).convert("RGB")
        high_img = Image.open(high_path).convert("RGB")

        low_img = self.transform(low_img)
        high_img = self.transform(high_img)

        return low_img, high_img

print("Dataset Class Ready")

# ============================================================
# STEP 6 : LOAD DATASET train particular images in batches ex 4
# ============================================================

train_dataset = LowLightDataset(
    LOW_LIGHT_PATH,
    NORMAL_LIGHT_PATH
)

train_loader = DataLoader(
    train_dataset,
    batch_size=8,
    shuffle=True
)

print("Total Training Images :", len(train_dataset))

# ============================================================
# STEP 7 : U-NET MODEL low light to enhance image
# ============================================================

class DoubleConv(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(out_channels, out_channels, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.conv(x)


class UNet(nn.Module):
    def __init__(self):
        super().__init__()

        self.down1 = DoubleConv(3, 64)
        self.pool1 = nn.MaxPool2d(2)

        self.down2 = DoubleConv(64, 128)
        self.pool2 = nn.MaxPool2d(2)

        self.middle = DoubleConv(128, 256)

        self.up1 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.conv1 = DoubleConv(256, 128)

        self.up2 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.conv2 = DoubleConv(128, 64)

        self.final = nn.Conv2d(64, 3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        d1 = self.down1(x)
        p1 = self.pool1(d1)

        d2 = self.down2(p1)
        p2 = self.pool2(d2)

        mid = self.middle(p2)

        u1 = self.up1(mid)
        u1 = torch.cat([u1, d2], dim=1)
        u1 = self.conv1(u1)

        u2 = self.up2(u1)
        u2 = torch.cat([u2, d1], dim=1)
        u2 = self.conv2(u2)

        out = self.final(u2)
        return self.sigmoid(out)

print("Model Class Ready")

# ============================================================
# STEP 8 : DEVICE + MODEL fake or real detects
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = UNet().to(device)

criterion = nn.L1Loss()
optimizer = optim.Adam(model.parameters(), lr=0.0001)

print("Using Device :", device)

# ============================================================
# STEP 9 : TRAINING (50 EPOCHS)
# ============================================================

EPOCHS = 50
best_loss = float("inf")

print("Training Started...")

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0

    loop = tqdm(train_loader)

    for low, high in loop:
        low = low.to(device)
        high = high.to(device)

        output = model(low)
        loss = criterion(output, high)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

        loop.set_description(f"Epoch [{epoch+1}/{EPOCHS}]")
        loop.set_postfix(loss=loss.item())

    avg_loss = total_loss / len(train_loader)

    print(f"Epoch {epoch+1} Average Loss : {avg_loss:.4f}")

    if avg_loss < best_loss:
        best_loss = avg_loss
        torch.save(model.state_dict(), SAVE_MODEL_PATH)
        print("Best Model Saved")

print("Training Completed")

# ============================================================
# STEP 10 : LOAD BEST MODEL optimizer
# ============================================================

model.load_state_dict(
    torch.load(
        SAVE_MODEL_PATH,
        map_location=device
    )
)

model.eval()

print("Best Model Loaded Successfully")

# ============================================================
# STEP 11 : TESTING + METRICS (10 IMAGES)
# ============================================================

def calculate_brightness(img):
    return np.mean(img)

def enhance_and_test(image_name):
    transform = transforms.Compose([
        transforms.Resize((128, 128)),
        transforms.ToTensor()
    ])

    low_path = os.path.join(TEST_LOW_PATH, image_name)
    high_path = os.path.join(TEST_HIGH_PATH, image_name)

    low_img = Image.open(low_path).convert("RGB")
    high_img = Image.open(high_path).convert("RGB")

    input_tensor = transform(low_img).unsqueeze(0).to(device)
    gt_tensor = transform(high_img).unsqueeze(0).to(device)

    with torch.no_grad():
        output_tensor = model(input_tensor)

    output_img = output_tensor.squeeze(0).cpu().permute(1,2,0).numpy()
    output_img = np.clip(output_img, 0, 1)

    low_np = transform(low_img).permute(1,2,0).numpy()
    high_np = transform(high_img).permute(1,2,0).numpy()

    psnr_value = peak_signal_noise_ratio(
        output_tensor.cpu(),
        gt_tensor.cpu()
    ).item()

    ssim_value = structural_similarity_index_measure(
        output_tensor.cpu(),
        gt_tensor.cpu()
    ).item()

    brightness_before = calculate_brightness(low_np)
    brightness_after = calculate_brightness(output_img)

    print("="*50)
    print("Image :", image_name)
    print(f"PSNR Score : {psnr_value:.2f}")
    print(f"SSIM Score : {ssim_value:.4f}")
    print(f"Brightness Before : {brightness_before:.4f}")
    print(f"Brightness After  : {brightness_after:.4f}")
    print("="*50)

    plt.figure(figsize=(15,5))

    plt.subplot(1,3,1)
    plt.imshow(low_np)
    plt.title("Low Light Input")
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.imshow(output_img)
    plt.title("Enhanced Output")
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.imshow(high_np)
    plt.title("Ground Truth")
    plt.axis("off")

    plt.show()

# ============================================================
# STEP 12 : TEST ON 10 IMAGES
# ============================================================

test_images = sorted(os.listdir(TEST_LOW_PATH))[:10]

for img in test_images:
    enhance_and_test(img)

print("Project Completed Successfully")
```
